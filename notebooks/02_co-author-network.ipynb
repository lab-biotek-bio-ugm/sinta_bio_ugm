# Co-author network
Main authors (red) → their Google Scholar co-authors (gray). The subgraph keeps only co-authors shared by more than one main author.

```python
import sys
sys.path.insert(0, '..')
from sinta_ls import FIGURES, PROCESSED
from sinta_ls import analysis as a

DATASET = "ALL"
G = a.coauthor_graph(a.load(DATASET))
H = a.connecting_subgraph(G)
G.number_of_nodes(), G.number_of_edges(), H.number_of_nodes()
```

```python
figs, graphs = FIGURES / DATASET, PROCESSED / DATASET
a.draw(G, figs / '02.0_co-author_graph.png')
a.draw(H, figs / '02.2_connecting_co-author_graph.png')
a.nx.write_graphml(G, graphs / 'co-author_network.graphml')
a.nx.write_graphml(H, graphs / 'co-author_subgraph_network.graphml')
```
